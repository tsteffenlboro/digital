## Define our ground truth

We pick $$(u-0.5)^2+1=u^2-u+1.25$$


```python
def real(u):
  return (u-0.5)**2+1

def dreal(u):
  return 2*(u-0.5)

u = np.array([-1,0,1])
# Note: we are processing the whole vector
# This requires very specific data structures
y = real(u)

print(u)
print(y)
```

## Using Derivatives

* Calculate the slope
* Calculate the slope change
* Find the parameters

```python

dy = y[1:]-y[:-1]
ddy = dy[1:]-dy[:-1]

c2 = np.mean(ddy)/2
c1 = np.mean(dy)
c0 = y[1]

print(c2,c1,c0)
```

# Using a Square Monomial

```python
def fb(u):
  return u**2
b = fb(u)
up = np.arange(-1, 1.1, 0.1)

plt.plot(up, fb(up), label='$u^2$')
plt.plot(up, 2*fb(up), label='$2u^2$')
plt.plot(up, 3*fb(up), label='$3u^2$')
plt.plot(u, y, 'x', label='y')
plt.legend()
```

## Single Basis Regression

Anyway, let us use the regression equation.



```python
c = np.dot(y,b)/np.dot(b,b)
print(c)

plt.plot(up, c*fb(up), label='$cu^2$')
plt.plot(u, y, 'x', label='y')
plt.legend()
```
